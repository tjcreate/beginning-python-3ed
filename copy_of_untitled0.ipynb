```python
# This creates and plots two Gaussian clusters in python
# One is centered at (0,0) and the other one at a random position at a unitary distance of the origin.
# Each sample has a label depending on its cluster

import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs
N_dimensions=2
center_1=np.zeros(N_dimensions)
center_2=np.random.randn(N_dimensions)
center_2=center_2/np.linalg.norm(center_2)

# Generate two Gaussian clusters
X, y = make_blobs(n_samples=200, centers=[center_1, center_2], cluster_std=0.25)
y=2*(y-1/2) # This transforms the labels to -1 and 1

# Plot the clusters
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.title('Put here your own title')
plt.xlabel('Name your Feature 2')
plt.ylabel('Name your Feature 1')
plt.show()
```
